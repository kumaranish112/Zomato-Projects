```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataframe = pd.read_csv("Zomato data .csv")
print(dataframe)
```

```python
dataframe
```

```python
# convert the data type of column - rate
```

```python
def handleRate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)
dataframe['rate'] = dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

```python
# Type of resturant
```

```python
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of restaurant")
```

```python
# conclusion - majority of the resturant falls in dining category
```

```python
dataframe.head()
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c="red", marker="o")
plt.xlabel("Type of restaurant", c="green", size=20)
plt.ylabel("Votes", c="green", size=20)
```

```python
# conclution - dining restaurant has received maximum votes
```

```python
dataframe.head()
```

```python
plt.hist(dataframe['rate'],bins =5)
plt.title("rating distribution")
plt.show()
```

```python
# conclution- the majority restarurant received rating from 3.5 to 4
```

```python
# Average order spending by couples
```

```python
dataframe.head()
```

```python
couple_data = dataframe['approx_cost(for two people)']
sns.countplot(x=couple_data)
```

```python
# CONCLUTION - THE MAJORITY OF COUPLES PREFRE RESTAURENT WITH AN APPROXIMATE COST OF TWO PEOPLES 
```

```python
# WITH MODE RECEIVES MAXIMUM RATING
```

```python
dataframe.head()
```

```python
plt.figure(figsize= (6,6))
sns.boxplot(x = 'online_order', y = 'rate', data= dataframe)
```

```python
# CONCLUTION - OFFLINE ORDER RECEIVED LOWER RATING IN COMPARISON TO ONLINE ORDER 
```

```python
dataframe.head()
```

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size',fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap="YlGnBu", fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (Type)")
plt.show()
```

```python
# CONCLUTION- DINING RESTAURANTS PRIMARILY ACCEPT OFFLINE ORDERS, WHEREAS CAFES PRIMARILY RECEIVE ONLINE ORDERS. THIS SUGGESTS THAT CLIENT
#             PREFER ORDER IN PERSON AT RESTAURANTS, BUT PREFER ONLINE ORDERIING AT CAFES. 
```
